```python
from transformers import pipeline

# Load a pretrained text-generation model
generator = pipeline(
    "text-generation",
    model="openai-community/gpt2"
)

print("GPT-2 loaded successfully!")
```

```python
# Start with a product description prompt
prompt = "The automatic umbrella helps busy people"

# Ask GPT-2 to continue the text
result = generator(
    prompt,
    max_new_tokens=90,
    do_sample=True,
    temperature=0.3,
    top_k=50,
    num_return_sequences=1,
    pad_token_id=generator.tokenizer.eos_token_id
)

print(result[0]["generated_text"])
```

```python
prompt = "A small business can use automatic umbrellas to"

result = generator(
    prompt,
    max_new_tokens=60,
    do_sample=True,
    temperature=0.8,
    pad_token_id=generator.tokenizer.eos_token_id
)

print(result[0]["generated_text"])
```
